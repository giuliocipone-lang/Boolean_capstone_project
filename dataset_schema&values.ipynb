## Schema dati del dataflow

Le dimensioni del dataset sono:
FREQ, ESITO, ITTER107, TIPO_DATO e SELECT_TIME

- Dimensione: FREQ
- id="CL_FREQ"
- URL: https://esploradati.istat.it/SDMXWS/rest/codelist/IT1/CL_FREQ
- Name: Frequenza
- Valori possibili:
-- ID	Descrizione
-- A	annuale
-- B	business (non supportato)
-- D	giornaliero
-- E	event (non supportato)
-- H	semestrale
-- M	mensile
-- Q	trimestrale
-- W	settimanale

- Dimensione: ESITO
- id="CL_ESITO"
- URL: https://esploradati.istat.it/SDMXWS/rest/codelist/IT1/CL_ESITO
- Name: Esito
- Valori possibili: <br>
M Morto <br>
F Ferito <br>
9 Totale
