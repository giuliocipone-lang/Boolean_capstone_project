## Bitcoin - prezzo ultimo anno

```python
url = 'https://api.coingecko.com/api/v3/coins/bitcoin/market_chart'
#ogni API ha documentato i parametri di cui ha bisogno
params = {
    'vs_currency': 'usd',
    'days': '365',
    'interval': 'daily'
}
```

```python
response = requests.get(url, params=params)
data = response.json()
data
#la risposta è una lista di liste di valori
```

```python
df_btc = pd.DataFrame(data['prices'], columns=['timestamp', 'price_usd'])
df_btc['date'] = pd.to_datetime(df_btc['timestamp'], unit='ms')
df_btc
```

```python
import requests
import time

url = "../DCIS_INCIDMORFER_COM.xml"
headers = {'Accept': 'application/vnd.sdmx.data+json;version=1.0.0'}

for attempt in range(15):
    try:
        response = requests.get(url, verify=False, timeout=30, headers=headers)
        print("Connessione riuscita!")
        print(response.status_code)
        break

    except requests.exceptions.RequestException:
        print(f"Tentativo {attempt + 1} fallito. Riprovo tra 30 secondi...")
        time.sleep(30)
```
