## Bitcoin - prezzo ultimo anno

```python
url = 'https://api.coingecko.com/api/v3/coins/bitcoin/market_chart'
#ogni API ha documentato i parametri di cui ha bisogno
params = {
    'vs_currency': 'usd',
    'days': '365',
    'interval': 'daily'
}
```

```python
response = requests.get(url, params=params)
data = response.json()
data
#la risposta è una lista di liste di valori
```

```python
df_btc = pd.DataFrame(data['prices'], columns=['timestamp', 'price_usd'])
df_btc['date'] = pd.to_datetime(df_btc['timestamp'], unit='ms')
df_btc
```

## Istat

```python
import requests
import time
from io import StringIO
```

```python
#esploriamo constraints del dataset

url =   "https://esploradati.istat.it/SDMXWS/rest/availableconstraint/41_983/all/all?mode=available"
#headers = {'Accept': 'application/vnd.sdmx.data+csv;version=1.0.0'}

while True:
    response_constraints = requests.get(url, verify=False)

    if response_constraints.status_code == 200:
        print("Richiesta riuscita!")
        break

    print(f"Status code: {response_constraints.status_code}. Riprovo tra 15 secondi...")
    time.sleep(15)
```

### Dati incidenti

```python
# schema URL https://esploradati.istat.it/SDMXWS/rest/data/flowRef/dim1.dim2.dim3.dim4.dim5/
# schema URL https://esploradati.istat.it/SDMXWS/rest/data/flowRef/.........?startPeriod=2016&endPeriod=2


url =   "https://esploradati.istat.it/SDMXWS/rest/data/41_983?startPeriod=2001&endPeriod=2024"
headers = {'Accept': 'application/vnd.sdmx.data+csv;version=1.0.0'}

while True:
    response_incidenti = requests.get(url, headers=headers, verify=False)

    if response_incidenti.status_code == 200:
        print("Richiesta riuscita!")
        break

    print(f"Status code: {response_incidenti.status_code}. Riprovo tra 15 secondi...")
    time.sleep(15)
```

```python
#converto la stringa response.text come se fosse un file CSV. Poi Panda la converte in un df.
df_incidenti = pd.read_csv(StringIO(response_incidenti.text))
```
